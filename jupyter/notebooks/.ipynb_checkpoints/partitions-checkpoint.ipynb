```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('read-partitions').getOrCreate()
```

```python
# Full snapshot history
spark.sql("SELECT * FROM nessie.default.static_data_partitioned.history").show()

# All data files and which partition they belong to
spark.sql("SELECT * FROM nessie.default.static_data_partitioned.files").show()

# Snapshots
spark.sql("SELECT * FROM nessie.default.static_data_partitioned.snapshots").show()
```
